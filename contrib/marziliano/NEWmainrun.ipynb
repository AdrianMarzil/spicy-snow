```python
from pathlib import Path
import sys
from os.path import expanduser
sys.path.append(expanduser('../../'))

import xarray as xr
xr.set_options(use_bottleneck=False)

from spicy_snow import retrieve_snow_depth
from spicy_snow.IO.user_dates import get_input_dates
import shapely

# Define base working directory (portable)
work_dir = Path.home() / "Desktop" / "SpicySnow"
work_dir.mkdir(parents=True, exist_ok=True)

# Output file
out_nc = Path.home() / "Desktop" / "BPR_snd2016.nc"

# Area
area = shapely.geometry.box(-106.9, 36.9, -106.4, 37.4)

# Dates
dates = get_input_dates("2016-06-01", "2015-11-01")

# Run retrieval
ds = retrieve_snow_depth(area, dates, work_dir=str(work_dir))

# Save
ds.to_netcdf(out_nc)
```
